## (d) Consider the following two portfolios:

Portfolio A had 60% invested in Target, 20% in Tesla and 20% in Alphabet at the beginning of each month.

Portfolio B had 60% invested in Tesla, 20% in Target and 20% in Alphabet at the beginning of each month.

```python
stocks = ["Target", "Tesla", "Alphabet"]
a = np.array([0.6, 0.2, 0.2])
b = np.array([0.2, 0.6, 0.2])
mu = rets[stocks].mean().values.round(4)
S = rets[stocks].cov().values.round(4)
```

Using just the average returns and variances of the three stocks, plus the covariances between the stocks (calculated in parts a, b and c), calculate the average returns and variances of these two portfolios, and the covariance between the two portfolios.

```python
# average returns
a @ mu
```

```python
b @ mu
```

```python
# variance
a @ S @ a
```

```python
b @ S @ b
```

```python
# covariance
a @ S @ b
```

Now on a spreadsheet calculate the realized returns of each of the portfolios for each month. Using these realized returns calculate the portfolio average returns, variances and the covariance directly, confirming your answers from above.

```python
rA = rets[stocks] @ a
rB = rets[stocks] @ b
```
